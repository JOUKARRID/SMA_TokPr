```python
import tiktoken
```

Importe la bibliothèque de tokenisation d'OpenAI.

```python
tokenizer = tiktoken.encoding_for_model("gpt-4o")
print(tokenizer.name)
```

Récupère le tokenizer utilisé par gpt-4o
Résultat attendu : o200k_base

```python
tokenizer = tiktoken.get_encoding("o200k_base")
```

exemple de tokenization

```python
prompt = "Vous êtes un expert en marketing digital. C'est quoi un digital marketer ? "
tokens = tokenizer.encode(prompt)
print(tokens)
print(len(tokens))
```

Affiche chaque token séparé par | , par exemple
Vous| être|s| un| expert|... Onvoit ainsi que les espaces sont collés au début des tokens et que certains mots sontcoupés.

```python
for token in tokens:
    t = tokenizer.decode_single_token_bytes(token).decode("utf-8") 
    print(t, end='|')
```

```python
from langchain_openai import ChatOpenAI
from dotenv.ipython import load_dotenv
```

```python
load_dotenv(override=True)
```

```python
llm = ChatOpenAI(model="gpt-4o", temperature=0)
```

```python
resp = llm.invoke(input=[{"role":"system","content":"Vous êtes un expert en marketing digital"},{"role":"user","content":"C'est quoi un digital marketer ?"}])
print(resp)
```
